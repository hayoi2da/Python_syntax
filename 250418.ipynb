```python
# str.format(), f-string

print("%d %5d %05d" % (123, 123, 123))
print("{0:d} {1:5d} {2:5d}".format(123, 123, 123)) # {} 안의 0, 1, 2 는 format() 함수 안의 값 중에서 0번째, 1번째, 2번째에 대응함.

print("%d%d%d" % (321, 321, 321))
print("%1d %1d %1d" % (321, 321, 321))
print("{2:d} {0:5d} {1:05d}".format(321, 321, 321))
# print("{3:d} {4:5d} {1:05d}".format(321, 321, 321)) - 오류
print("{1:d} / {0:d} = {2:d}".format(3, 5, 15))

# > 오른쪽 정렬

print("{:>5}".format(123)) # 5자리 확보 후 오른쪽 정렬
print("{:1>5}".format(123)) # 5자리 확보 후 빈 자리에 1로 채움
print("{:^>5}".format(123)) # 5자리 확보 후 빈 자리에 ^로 채움

# < 왼쪽 정렬

print(f"{123:<5}") # 5자리 확보 후 왼쪽 정렬
print(f"{123:1<5}") # 5자리 확보 후 빈 자리에 1로 채움
print(f"{123:^^5}") # 5자리 확보 후 양쪽 빈 자리에 ^로 채움
```

```python
# 이스케이프 문자

# \n - 새로운 줄로 이동, 개행 : enter를 누른 효과
# \t - 다음 탭으로 이동 : Tab을 누른 효과
# \b - 뒤로 한 칸 이동 : Backspace를 누른 효과 - 위치에 따라 글자가 하나 지워짐.
# \\ - \ 출력
# \' - ' 출력
# \" - " 출력

print("한 행입니다. 또 한 행입니다.")
print("한 행입니다. \n또 한 행입니다.")
print("한 행입니다. \n 또 한 행입니다.")
print("한 행입니다.\n또 한 행입니다.")

print("니가 누구라고?\t하영입니다옹 ")
print("니가 누구라고? \n \t 하영입니다옹 ")

print("니가 누구라고?\b 하영입니다옹") # \b의 위치에 따라 ?가 지워짐
print("니가 누구라고?\b 하영입니다옹\b!") # \b의 위치에 따라서 ? 와 옹이 지워지고 옹은 -> !로 바뀜
print("니가 누구라고? \n하영입니다옹\b")

print("나는 지금뿌링뿌링 세트\\마라샹궈를 먹고 싶다")

print("my name is \" hayoi")
print("my name is \' hayoi")
# print("my name is \"" hayoi") - 오류
# print(\"my name is hayoi) - 오류
print("my name is \"hayoi\"")
print("my name is \' hayoi \'")

print("\n줄바꿈\n연습")
print("탭키\t연습")
print("\t탭키\t연습")
# \t는 몇 칸?	현재 커서 위치 기준으로 다음 8의 배수 위치까지 이동
# 한글은 왜 정렬 이상해짐?	한글은 보통 2칸 너비로 표시돼서 탭 정렬이 안 맞을 수 있음

# 첫 번째 \t: 커서를 0 → 8칸으로 이동
# → 여기에 "탭키" 출력 (한글이라 2칸씩, 총 4칸)

# 두 번째 \t: 커서가 현재 12칸쯤 위치에 있음 (8칸 + 한글 4칸)
# → 다음 8의 배수는 16칸 → 연습은 16칸 위치에 출력됨

# 그래서 체감되는 건?
# 첫 번째 \t는 시작 위치라 전체 8칸을 비워주고, 탭키가 와서 들여쓰기가 커보임
# 두 번째 \t는 이미 글자가 차 있어서 실제로는 짧게만 띄워진다
# 결과적으로 첫 번째 탭이 더 긴 것처럼 보이는 착시가 생김

print("탭키\t연습\t")
print("글자가 \"강조\" 되는 효과1")
print("글자가 \'강조\' 되는 효과2")
print("\\\\\\ 역슬래시 세 개 출력")
print(r"\n \t \" \\를 그대로 출력")
```

```python
# 변수의 선언

f = True # 불형 변수
u = 20251772 # 정수형 변수
c = 2.18 # 실수형 변수
k = "ㄹㅇㅋㅋ" # 문자열 변수
q = "zzz"

print(f)
print(u)
print(c)
print(k)
print(q)

print(type(f))
print(type(u))
print(type(c))
print(type(k))

f, u, c, k = True, 20251772, 2.18, "ㄹㅇㅋㅋ"

print(f,u,c,k)

# 변수에 변수를 대입하는 방식, 숫자끼리 연산한 결과를 대입하는 방식, 변수와 숫자를 연산한 결과를 대입하는 방식
f = 100
u = f
c = 100 + f
k = f / f
q = int(f/f)
print(f, u, c, k, q)
```

```python
# 연속된 값을 대입하는 방식 - =는 맨 뒤부터 처리됨.

f = u = c = k = 20251772

print(f, u, c, k)

f = 20251772
u = f
c = u
k = c

print(f, u, c, k)

# 연산 결과를 자신의 값에 다시 대입하는 방식, 자신의 값을 누적할 경우 사용
# 주의할 점 - 처음에 변수에 0 또는 다른 값이 들어 있어야 함.

# fuck = fuck + 100 - NameError: name 'fuck' is not defined, fuck = 0 또는 다른 값을 대입시켜 놓아야 함.
# print(fuck)

fuck = 10
fuck = fuck + 100 # fuck += 100 이것도 됨

print(fuck)

fuck = 20251772
print(type(fuck))
fuck = "정하영"
print(type(fuck))
fuck = 2.18
print(type(fuck))
```
