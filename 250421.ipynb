```python
# 진수 변환 - int('숫자', 진수)를 사용하면 각 진수를 10진수로 변환한다.

# 2진수 - 앞에 0b 나 0B 를 붙임

print(0b10010011)
print(int('10010011',2))

# 16진수 - 앞에 0x 나 0X 를 붙이고, int() 함수를 사용

print(0x13)
print(int('13',16))
print(0x93, int('93',16))
# print(0x93; int('93',16)) - 왜 오류 뜸? >> ; 두 개의 명령을 구분, 함수 괄호 안에는 쓸 수 없음.
print(0x93); print(int('93',16))

# 8진수 - 앞에 0o 나 0O를 붙임.
```

```python
# 진수 변환 2 - hex(숫자) >> 16진수로, oct() >> 8진수로, bin() >> 2진수로

print(bin(11), bin(0o11), bin(0x11))
print(oct(11), oct(0b11), oct(0x11))
print(hex(11), hex(0b11), hex(0o11))
```

```python
# 진수 변환 프로그램 - 오류가 있음, 16진수인 문자열을 받지 못함.

jinsu = int(input("입력 진수 결정(16/10/8/2):"))
value = int(input("값 입력:")) # 내가 입력한 것

if jinsu == 16:
    print(f"16진수 ==> {hex(value)}")
if jinsu == 10:
    print(f"10진수 ==> {value}")
if jinsu == 8:
    print(f"8진수 ===> {oct(value)}")
if jinsu == 2:
    print(f"2진수 ==> {bin(value)}")
```

```python
# 진수 변환 프로그램 모범 답안 - 16진수는 알파벳을 출력하므로 값을 받는 변수에는 문자열로 입력 받은 후 10진수로 변환시켜야 함.

jinsu = int(input("입력 진수 결정(16/10/8/2):"))
value = input("값 입력:") # 16진수는 알파벳(문자열)을 출력하므로 int() 사용 x

value10 = 0

# 원래 진수 변환 - int('숫자', 진수) -> 여기서는 int(변수, 진수)
# 어떤 진수든 그 진수에 맞는 숫자 문자만 써야 함.
#'숫자처럼 보이는 문자열'을 숫자로 바꿔달라고 할 때는, 그 문자열이 몇 진수로 쓰였는지 알려줘야 해! 그래서 그때 base가 필요한 거야.

# int()에 넣는 값 | base가 필요한가? | 왜?
# '1010'(문자열) | ✅ 필요함 | "이 문자열이 몇 진수인지" 알려줘야 해
# 1010(숫자) | ❌ 필요 없음 (무시됨) | 숫자는 이미 정수니까 변환할 필요 없음

if jinsu == 16:
    value10 = int(value, 16)
if jinsu == 10:
    value10 = int(value, 10)
if jinsu == 8:
    value10 = int(value, 8)
if jinsu == 2:
    value10 = int(value, 2)

print("16진수 ==>", hex(value10))
print("10진수 ==>", (value10))
print("8진수 ==>", oct(value10))
print("2진수 ==>", bin(value10))
```

```python
# 진수 변환 프로그램 모범 답안 - 16진수는 알파벳을 출력하므로 값을 받는 변수에는 문자열로 입력 받은 후 10진수로 변환시켜야 함.

jinsu = int(input("입력 진수 결정(16/10/8/2):"))


if jinsu == 16 or jinsu == 10 or jinsu == 8 or jinsu == 2:
    value = input("값 입력:") # 16진수는 알파벳(문자열)을 출력하므로 int() 사용 x
    value10 = 0
    if jinsu == 16:
        value10 = int(value, 16)
    if jinsu == 10:
        value10 = int(value, 10)
    if jinsu == 8:
        value10 = int(value, 8)
    if jinsu == 2:
        value10 = int(value, 2)

    print("16진수 ==>", hex(value10))
    print("10진수 ==>", (value10))
    print("8진수 ==>", oct(value10))
    print("2진수 ==>", bin(value10))

else:
    print("16, 10, 8, 2 숫자 중 하나만 입력하세요.")
```

```python
# main() 함수의 효과를 보여주는 방법 -> 파이썬은 main() 함수가 없음

## 함수 선언 부분 ##
def myFunc():
    print('함수를 호출함.')

## 전역 변수 선언 부분 ##
gVar = 100

## 메인 코드 부분 ##
if __name__=='__main__':
    print('메인 함수 부분이 실행됩니다')
    myFunc()
    print('전역 변수 값:', gVar)
```

```python
# main() 함수의 효과를 보여주는 방법 -> 파이썬은 main() 함수가 없음

## 함수 선언 부분 ##
def myFunc():
    print('함수를 호출함.')

## 전역 변수 선언 부분 ##
gVar = 100

## 메인 코드 부분 ## - 수정본
def main():
    print("메인 함수 부분이 실행됩니다.")
    myFunc()
    print('전역 변수 값:', gVar)

if __name__=='__main__':
    main()
```
