```python
# print() 함수를 사용한 깔끔한 출력

# %[채움문자][정렬/부호][너비].[소수점 자리수]형식문자

# %d, %x, %o - 정수(10진수, 16진수, 8진수)
# %f - 실수(소수점이 붙은 수)
# %c - 한 글자
# %s - 두 글자 이상인 문자열

# 정수형 데이터 서식
print("%d" % 123)
print("%5d" % 123) # 5자리 확보 후 오른쪽에 붙여서 정렬
print("%05d" % 123) # 5자리 확보 후 오른쪽에 불여서 정렬 + 빈칸을 0으로 채움

# 실수형 데이터 서식
print("%f" % 123.45) # 소수점 아래 여섯 자리까지 무조건 출력
print("%7.1f" % 123.45) # 7자리 확보 후 소수점 아래 첫째 자리만 출력, 소수점 아래 둘째 자리에서 반올림 / %.1f → 소수점 첫째 자리까지 반올림
print("%7.3f" % 123.45) # 7자리 확보 후 소수점 아래 셋째 자리까지 출력, 오른쪽 빈칸은 0으로 채움

# 문자열 데이터 서식
print("%s" % "Python") # 자릿수만큼 출력
print("%10s" % "Python") # 10자리 확보 후 오른쪽 정렬
```

```python
# 헷갈리는 부분

print("%1.5d" % 123) # C 스타일 포맷(%)에서는 '0' 이외의 다른 문자로는 채울 수 없음
# 파이썬의 str.format() 또는 f-string을 사용하면 더 유연한 포맷팅이 가능하고, 0 외의 문자로도 채울 수 있음

# C 스타일
print("%05d" % 123)  # 00123

# str.format()
print("{:05d}".format(123))  # 00123

# f-string
print(f"{123:05d}")  # 00123


print("%f" % 123)
print("%7.4f" % 123.45555)
```

```python
# chat gpt 실습

print("{:*>5}".format(123)) # ****123 중 5자리 확보
print(f"{123:*<5}") # 왼쪽 정렬, 오른쪽을 *로 채움
print(f"{123:*^5}") # 가운데 정렬, 양쪽을 *로 채움
```
