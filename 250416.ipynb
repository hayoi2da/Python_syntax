```python
# print() 함수의 서식
# 서식의 개수와 따옴표 뒤에 나오는 숫자(또는 문자)의 개수는 같아야 함.
# %d는 정수(Decimal) 의미

"%d" % 100 # print 함수가 없었을 때는 하나의 값이 출력됐음.
'%d' % 100 # %d는 정수(Decimal) 의미

print("100") # 문자열 100 출력
print("%d" % 100) # 숫자 100 의미

print("100 + 100") # 문자열 100 + 100 출력
print("%d" % (100 + 100)) # 숫자 100과 숫자 100을 더한 값 200 출력

# print("%d" % (100, 200)) - 오류 코드 : 서식의 개수와 숫자의 개수가 같아야 함.
print("%d %d" % (100, 200))

# print("%d %d" % 100) - 오류 코드 : 서식의 개수와 숫자의 개수가 같아야 함.
print("%d" % 100)
```

```python
# print() 함수를 사용한 다양한 출력

print("%d / %d = %d" % (100, 200, 0.5)) # 서식에 정수형(%d)을 사용했기 때문에 0.5(실수)가 0(정수)으로 출력

# %d, %x, %o - 정수(10진수, 16진수, 8진수)
# %f - 실수(소수점이 붙은 수)
# %c - 한 글자
# %s - 두 글자 이상인 문자열

# 수정
print("%d / %d = %f" % (100, 200, 0.5)) # 0이 너무 많다.

# 수정정
print("%d / %d = %5.1f" % (100, 200, 0.5))
```
